```python
def fizzbuzz(n):
    for i in range(1,n+1):
        if i%3==0 and i%5==0:
            print("Fizzbuzz",end=" ")
        elif i%3==0:
            print("Fizz",end=" ")
        elif i%5==0:
            print("buzz",end=" ")
        else:
            print(i,end=" ")
fizzbuzz(15)
```

```python
def is_prime(n):
    if n <= 0 or n==1:
        return False
    for i in range(2,n):
        if n%i==0:
            print(f"{n} is not a prime number") 
            return
    print(f"{n} is a prime number") 
is_prime(9)
```

```python
def is_factorial(n):
    result = 1
    if n<=0 or n==1:
        return 1
    for i in range(2,n+1):
        result *= i
    print(f"{result} is a factorial for:",i) 
is_factorial(12)
```

```python
def fibonacci(n):
    a, b = 0, 1
    result = []

    for _ in range(n):
        result.append(a)
        a, b = b, a + b
    return result
fibonacci(18)
```

```python
def reverse_string(nums):
    result = ""
    for i in nums:
        result = i + result
    return result
reverse_string("shiva")
```

```python
def reverse_string(s):
    result = ""  
    for char in s:
        result = char + result    
    return result  

(reverse_string("SHIVA"))
```

```python
def reverse_numeric_list(nums):
    result = [] 
    for i in range(len(nums) - 1, -1, -1):
        result.append(nums[i])  
    return result

(reverse_numeric_list([10, 20, 30, 40, 50]))
```

```python
def countvowels(n):
    vowels, count = "aeiou", 0
    for i in n.lower():
        if i in vowels:
            count += 1
    return count
countvowels("shivA")
```

```python
def maxmin(lst):
    if len(lst)==0 :
        return None
    
    max_val = lst[0]
    min_val = lst[0]

    for i in lst:
        if i > max_val:
            max_val = i
        if i < min_val:
            min_val = i
    return max_val, min_val
maxmin([1,2,3,4,5,6,7])
```

```python
def remove_duplicates(listt):
    result = []
    for i in listt:
        if i not in result:
            result.append(i)
    return result
remove_duplicates([1,2,2,2,2,3,3,4,5,6,6])
```

```python
d1 = {"a": 1, "b": 2}
d2 = {"b": 20, "c": 3}
def merge_dicts(d1, d2):
    result = d1.copy()
    result.update(d2)
    return result
print(merge_dicts(d1, d2))
```

```python
def word_frequency(sentence):
    freq = {}

    for word in sentence.lower().split():
        freq[word] = freq.get(word, 0) + 1

    return freq
word_frequency('My name is shiva and my location is hyderabad')
```

```python
def is_palindrome(n):
    clean = ""
    for i in n:
        if i.isalnum():
            clean += i.lower()
    return clean == clean[::-1]
is_palindrome("A man a plan a canal Panama")
```

```python
def is_palindrome(s):
    # Step 1: Filter out only alphanumeric characters and make it lower case
    clean_s = "".join([char.lower() for char in s if char.isalnum()])
    
    # Step 2: Compare with its reverse sequence
    return clean_s == clean_s[::-1]

print(is_palindrome("A man a plan a canal Panama"))
```

```python
def sum_digits(n):
    str_n = str(abs(n))
    result = 0
    for i in str_n:
        result += int(i)
    return result
sum_digits(-16782)
```

```python
def count_occurences(lst):
    result = {}
    for i in lst:
        if i not in result:
            result[i] = 1
        else:
            result[i] += 1
    return result
count_occurences(['a', 'b', 'a', 'c', 'b', 'a'])
```

```python
def flatten(n):
    result = []
    for i in n:
        for j in i:
            result.append(j)
    return result
flatten([[1, 2], [3, 4], [5],[5]])
```

```python
def second_larget(lst):
    if len(lst) < 2:
        return None
    largest = float('-inf')
    second_largest = float('inf')
    for i in lst:
        if i > largest:
            second_largest = largest
            largest = i
        elif i > second_largest and i != largest:
            second_largest = i
    return second_largest if second_largest != float('-inf') else None
second_larget([7,7,1])
```

```python
def is_anagram(s1,s2):
    if len(s1) != len(s2):
        return False
    count_s1 = {}
    count_s2 = {}
    for i in count_s1:
        if i not in count_s1:
            count_s1 = 1
        else:
            count_s1 += 1
    for j in count_s2:
        if j not in count_s2:
            count_s2 = 1
        else:
            count_s2 += 1
    return count_s1 == count_s2
print(is_anagram("apple", "pale"))
```

```python
def single_multiplication_table(n):
    # Loop runs from 1 exactly up to n (so we use n + 1 as stop boundary)
    for i in range(1, 11):
        product = n * i
        # Standard clean alignment template format configuration
        print(f"{n} x {i} = {product}")

single_multiplication_table(7)
```

```python
def intersection(l1, l2):
    set1 = set(l1)
    set2 = set(l2)
    
    common_elements = set1 & set2
    return sorted(list(common_elements))

def union(l1, l2):
    set1 = set(l1)
    set2 = set(l2)
    
    # Step 2: Use the bitwise OR operator (|) to combine all unique elements
    all_elements = set1 | set2
    return sorted(list(all_elements))

list_a = [4, 1, 2, 2]
list_b = [2, 2, 3]

print("Intersection:", intersection(list_a, list_b)) 
print("Union:       ", union(list_a, list_b))        
```

```python
def sort_by_value(d):
    # Step 1: Extract dictionary items as a list of (key, value) tuples
    dict_items = list(d.items())
    
    # Step 2: Use a custom tracking function to tell sorted() to look at the value (index 1)
    def get_value(item):
        return item[1]
    
    # Step 3: Sort the tuples based on the values
    sorted_tuples = sorted(dict_items, key=get_value)
    
    # Step 4: Convert the sorted list of tuples back into a structured dictionary
    return dict(sorted_tuples)

# Test Execution Matrix
data = {"b": 2, "a": 1, "c": 3}
print("Original: ", data)
print("Sorted:   ", sort_by_value(data))  # Expected Output: {"a": 1, "b": 2, "c": 3}
```
