```python
def fizzbuzz(n):
    for i in range(1,n+1):
        if i%3==0 and i%5==0:
            print("Fizzbuzz",end=" ")
        elif i%3==0:
            print("Fizz",end=" ")
        elif i%5==0:
            print("buzz",end=" ")
        else:
            print(i,end=" ")
fizzbuzz(15)
```

```python
def is_prime(n):
    if n <= 0 or n==1:
        return False
    for i in range(2,n):
        if n%i==0:
            print(f"{n} is not a prime number") 
            return
    print(f"{n} is a prime number") 
is_prime(9)
```

```python
def is_factorial(n):
    result = 1
    if n<=0 or n==1:
        return 1
    for i in range(2,n+1):
        result *= i
    print(f"{result} is a factorial for:",i) 
is_factorial(12)
```

```python
def fibonacci(n):
    a, b = 0, 1
    result = []

    for _ in range(n):
        result.append(a)
        a, b = b, a + b
    return result
fibonacci(18)
```

```python
def reverse_string(nums):
    result = ""
    for i in nums:
        result = i + result
    return result
reverse_string("shiva")
```

```python
def reverse_string(s):
    result = ""  
    for char in s:
        result = char + result    
    return result  

(reverse_string("SHIVA"))
```

```python
def reverse_numeric_list(nums):
    result = [] 
    for i in range(len(nums) - 1, -1, -1):
        result.append(nums[i])  
    return result

(reverse_numeric_list([10, 20, 30, 40, 50]))
```

```python
def countvowels(n):
    vowels, count = "aeiou", 0
    for i in n.lower():
        if i in vowels:
            count += 1
    return count
countvowels("shivA")
```

```python
def maxmin(lst):
    if len(lst)==0 :
        return None
    
    max_val = lst[0]
    min_val = lst[0]

    for i in lst:
        if i > max_val:
            max_val = i
        if i < min_val:
            min_val = i
    return max_val, min_val
maxmin([1,2,3,4,5,6,7])
```

```python
def remove_duplicates(listt):
    result = []
    for i in listt:
        if i not in result:
            result.append(i)
    return result
remove_duplicates([1,2,2,2,2,3,3,4,5,6,6])
```

```python
d1 = {"a": 1, "b": 2}
d2 = {"b": 20, "c": 3}
def merge_dicts(d1, d2):
    result = d1.copy()
    result.update(d2)
    return result
print(merge_dicts(d1, d2))
```

```python
def word_frequency(sentence):
    freq = {}

    for word in sentence.lower().split():
        freq[word] = freq.get(word, 0) + 1

    return freq
word_frequency('My name is shiva and my location is hyderabad')
```
