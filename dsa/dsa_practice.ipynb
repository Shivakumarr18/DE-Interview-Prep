```python
def maxwealth(accounts):
    max_wealth = 0
    for row in accounts:
        row_sum = sum(row)
        if row_sum > max_wealth:
            max_wealth = row_sum
    return max_wealth
maxwealth([[1,2],[1,3],[1,2,3],[1,5,6]])

#Time and space complexity for above problem:

# Time is O(m × n) because I visit every cell in the grid. Space is O(1) because I only use a fixed number of variables regardless of input size.

# PROBLEM 1 — Richest Customer Wealth
# Pattern  : 2D array traversal + row aggregation
# Time     : O(m × n) — visit every cell once
# Space    : O(1) — only a max variable
# Key steps: 1. Loop through each row
#            2. Sum the row
#            3. Track maximum sum seen so far
# Special  : sum() on a row is O(n) — not O(1)
```

```python
def twosums(nums,target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return[seen[complement],i]
        seen[num] = i
print(twosums([2, 7, 11, 15], 9))   
print(twosums([3, 2, 4], 6))         
print(twosums([3, 3, 1],4))

#"Time complexity is O(n) — we make a single pass through the array and each hash map lookup costs O(1). Space complexity is O(n) — in the worst case we 
# store every element in the hash map before finding the answer. This is a deliberate tradeoff — we traded O(n) space to bring time down from O(n²) to O(n).

# PROBLEM 2 — Two Sum
# Pattern  : Complement + Hash Map
# Time     : O(n) — single pass, O(1) lookup each step
# Space    : O(n) — hash map stores at most n elements
# Key steps: 1. For each num, calculate complement = target - num
#            2. Check if complement already in hash map
#            3. If yes → return [seen[complement], i]
#            4. If no → store seen[num] = i
# Special  : Check BEFORE storing — avoids matching with itself
# Tradeoff : Bought O(n) time by spending O(n) space
```

```python
def negativecount(nums):
    count = 0
    for i in nums:
        for j in i:
            if j<0:
                count += 1
    return count
negativecount([[4,3,2,-1],[3,2,1,-1],[1,1,-1,-2],[-1,-1,-2,-3]])

#This is the BruteForce Solution. Time and space complexity is : O(mxn). 
```

```python
def first_negative_index(row):
    left = 0
    right = len(row)

    while left < right:
        mid = (left+right)//2
        if row[mid] >= 0:
            left = mid + 1
        else:
            right = mid
    return left

def negativecount(nums):
    count = 0
    n = len(nums[0])
    for row in nums:
        j = first_negative_index(row)
        count += n-j
    return count

print(negativecount([[4,3,2,-1],[3,2,1,-1],[1,1,-1,-2],[-1,-1,-2,-3]]))
print(negativecount([[3,2],[1,0]]))
print(negativecount([[-1]]))

# Time complexity is O(m × log n) — we iterate through m rows, and for each row
# we run binary search which costs O(log n) by halving the search space each time.
# Space complexity is O(1) — we only store left, right, mid, and count variables
# regardless of input size. No extra structure that grows with input.
# This is a deliberate upgrade from brute force O(m × n) — we exploited the sorted
# property of each row to skip cells instead of visiting every one.

# PROBLEM 3 — Count Negative Numbers in Sorted Matrix
# Pattern  : Sorted array → Binary Search per row
# Time     : O(m × log n) — binary search each of m rows
# Space    : O(1) — only index variables
# Key steps: 1. For each row, binary search for first negative index j
#            2. Count += n - j (everything from j onwards is negative)
#            3. right = len(row) not len-1 — handles all-positive row
# Special  : Brute force O(m×n) also valid — know both
# Tradeoff : Exploited sorted property to skip cells
```

```python
def nextPermutation(nums: list[int]):
    n = len(nums)
    i = n - 2
    
    while i >= 0 and nums[i] >= nums[i + 1]:
        i -= 1
        
    if i >= 0:
        j = n - 1
        while nums[j] <= nums[i]:
            j -= 1
        nums[i], nums[j] = nums[j], nums[i]
        
    left = i + 1
    right = n - 1
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1


my_nums = [2, 3, 5, 4, 1]

# 3. Call the function (It modifies the list in-place)
nextPermutation(my_nums)
print(my_nums)

# Time complexity is O(n) — three linear passes: pivot scan,
# swap candidate scan, and suffix reverse. Each touches at most n elements.
# Space complexity is O(1) — in-place modification using only
# index variables i, j, left, right. No extra data structure created.
# Pattern: right-to-left scan + swap + reverse suffix.

# Problem 4 — Next Permutation

# Pattern  : Right-to-left scan + swap + reverse
# Time     : O(n)
# Space    : O(1)
# Key steps: 1. Find pivot (first i where nums[i] < nums[i+1])
#            2. Find smallest number right of pivot greater than pivot
#            3. Swap them
#            4. Reverse everything after pivot index
# Special  : No pivot found → reverse entire array
```

```python
def ArrayGCD(nums: list[int]):
    min_num = min(nums)
    max_num = max(nums)

    a , b = min_num , max_num
    while b > 0:
        a , b = b , a % b
    return a
ArrayGCD([5,6,7,8])

# Time complexity is O(n) — finding max and min requires a full scan
# of the array. The Euclidean GCD computation on two numbers runs in
# O(log(min(a,b))) which is negligible compared to the array scan.
# Space complexity is O(1) — only fixed variables a, b used.
# No extra data structure created regardless of input size.

# Problem 5 — Find GCD of Array

# Pattern  : Math — Euclidean Algorithm
# Time     : O(n)
# Space    : O(1)
# Key steps: 1. Find max and min of array
#            2. Apply Euclidean Algorithm on just those two numbers
#            3. While b > 0: a, b = b, a % b
#            4. Return a when b reaches 0
# Special  : GCD of entire array not needed — only max and min
#            a, b = b, a % b is the Pythonic one-liner for Euclidean
```

```python
def isSelfDividing(num: int) -> bool:
    temp = num
    while temp > 0:
        digit = temp % 10
        if digit == 0 or num % digit != 0:
            return False
        temp = temp // 10
    return True
isSelfDividing(66)

# Time complexity is O(N × D) — N numbers in range, each checked
# against D digits. D ≤ 5 for constraint 10^4, making this
# effectively O(N) in practice.
# Space complexity is O(1) auxiliary — only temp, digit variables.
# Output list not counted as extra space (it's the required return value).
# Pattern: digit extraction using % 10 (last digit) and // 10 (shrink number).

# Problem 6 — Self Dividing Numbers

# Pattern  : Digit extraction + divisibility check
# Time     : O(N × D) → effectively O(N) since D ≤ 5
# Space    : O(1) auxiliary
# Key steps: 1. For each number in range
#            2. Extract each digit using % 10
#            3. Check digit != 0 AND num % digit == 0
#            4. Shrink number using // 10
#            5. If all digits pass → add to result
# Special  : Check digit == 0 BEFORE modulo — prevents ZeroDivisionError
#            Use temp variable — preserves original num for divisibility check
```
